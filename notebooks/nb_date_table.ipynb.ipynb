```python
from pyspark.sql import functions as F

gld_date = (
    spark.sql("""
        SELECT explode(
            sequence(
                to_date('2023-01-01'),
                to_date('2026-12-31'),
                interval 1 day
            )
        ) AS Date
    """)
    .withColumn("Year", F.year("Date"))
    .withColumn("MonthNumber", F.month("Date"))
    .withColumn("MonthName", F.date_format("Date", "MMMM"))
    .withColumn("MonthShortName", F.date_format("Date", "MMM"))
    .withColumn("YearMonth", F.date_format("Date", "yyyy-MM"))
    .withColumn("YearMonthSort", F.date_format("Date", "yyyyMM").cast("int"))
    .withColumn("QuarterNumber", F.quarter("Date"))
    .withColumn(
        "Quarter",
        F.concat(
            F.lit("Q"),
            F.quarter("Date")
        )
    )
    .withColumn("MonthStart", F.trunc("Date", "month"))
    .withColumn("MonthEnd", F.last_day("Date"))
    .withColumn("DayOfWeekNumber", F.dayofweek("Date"))
    .withColumn("DayName", F.date_format("Date", "EEEE"))
)
```

```python
(
    gld_date.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("gld_date")
)

print("Gold date dimension created successfully.")
```
